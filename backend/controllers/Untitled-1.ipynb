```python
import pandas as pd

file_path = r"C:\Users\TilaK\Downloads\W3.xlsx"
df = pd.read_excel(file_path, sheet_name="Sheet1")

def fill_blanks(series):
    result = []
    block = []
    for val in series:
        if pd.isna(val):  # blank cell
            result.append(", ".join(block))
        else:
            block.append(str(val))
            result.append(val)
    return result

df["IO NO"] = fill_blanks(df["IO NO"])

output_path = r"C:\Users\TilaK\Downloads\W3_processed_block.xlsx"
df.to_excel(output_path, sheet_name="Sheet1", index=False)

print("Processed file saved as:", output_path)
```
